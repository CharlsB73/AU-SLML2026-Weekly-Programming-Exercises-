```python
import numpy as np
from numpy.linalg import eig
```

```python
# 1) Obtain S
x = [[-1, -0.5],
        [-1,1],
        [0, 0.5],
        [0.5, 1.5],
        [1, 1]]

cov_matrix = np.cov(np.transpose(x))
print(cov_matrix)
# 2) Obtain eigenvector with highest eigenvalues

eigenvalues,eigenvectors=np.linalg.eigh(cov_matrix)
print('E-value:', eigenvalues)
print('E-vector', eigenvectors)

max_index = np.argmax(eigenvalues)

max_eigenvalue = eigenvalues[max_index]
max_eigenvector = eigenvectors[:, max_index]
print("Max eigenvector", max_eigenvector)
eigenvectors[0].T @ cov_matrix @ eigenvectors[0]

# print(max_eigenvector.T @ cov_matrix @ max_eigenvector)


```
