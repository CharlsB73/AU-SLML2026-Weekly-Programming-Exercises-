### 1) Estimate the mean of the Gaussian distribution, µM L, from the available data points by using the maximum likelihood (ML) method. (Score: 15%)

```python
import numpy as np
```

```python
x = [0.0557, 1.8529, 1.9311, 2.1905, 3.8884, 4.4384]
x_data = np.array(x)
N = len(x_data)
mu_mle = sum(x_data)/N
mu_mle
```

### 2) Estimate the variance of the Gaussian distribution. σ2 M L, from the available data points by using the maximum likelihood (ML) method

```python
dist = x_data - mu_mle
sqrd_dist = np.power(dist, 2)
var_mle = sum(sqrd_dist)/N
var_mle
```
