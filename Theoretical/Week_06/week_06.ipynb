```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
phi = [[1, -1],
       [1, 1],
       [1, 2]]
phi= np.array(phi)
phi_final = np.linalg.inv(phi.T @ phi) @ phi.T
T = [-1, 1, 2]
X = [1, 1.5, 3]
weights = phi_final @ X
print(weights)
f = []
for i in range(len(T)):
    temp = weights[0] + weights[1]*T[i]
    f.append(temp)
# 1st prediction
p1 = weights[0] + weights[1]*0
p2 = weights[0] + weights[1]*3
T2 = [0, 3]
X2 = [p1, p2]
plt.plot(T, f)
plt.scatter(T, X, color='blue', label='Data Points')
plt.scatter(T2, X2, color='red', marker='x')
plt.show()
```
